Here you don't have a collaborator to consult, but as the data is publicly available, google will help you answer any questions that may come up.

I just want to stress that this process of data visualisation and cleaning is laborious, and can take time - but it is also where making the right choices can make a huge difference in practice - often much more so than in the later, perhaps more interesting part where you are designing a model.
